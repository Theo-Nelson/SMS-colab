## Basecalling

Let us run download slow5-Dorado, which is a fork of ONT's Dorado basecaller with S/BLOW5 support

```python
! wget https://github.com/hiruna72/slow5-dorado/releases/download/v0.8.3/slow5-dorado-v0.8.3-x86_64-linux.tar.xz -O slow5-dorado.tar.xz && tar xf slow5-dorado.tar.xz
```

Now download the right basecalling model. Example dataset we downloaded is from the latest R10.4.1 chemistry, so we will get the latest (v5.0.0) and most accurate (super accuracy) model.

```python
! slow5-dorado/bin/slow5-dorado download --model dna_r10.4.1_e8.2_400bps_sup@v5.0.0
```

Now basecall the BLOW5 into a FASTQ file and see if the output looks good

```python
! slow5-dorado/bin/slow5-dorado basecaller dna_r10.4.1_e8.2_400bps_sup@v5.0.0 chr22_20240000_20290000.blow5 --emit-fastq > reads.fastq
```

```python
! head reads.fastq
```

```python
! wc -l reads.fastq
```

Aligning basecalled reads to the reference

Setup minimap2 aligner

```python
! wget https://github.com/lh3/minimap2/releases/download/v2.28/minimap2-2.28_x64-linux.tar.bz2 && tar xf minimap2-2.28_x64-linux.tar.bz2
```

Download the reference, in this case chr22 of the hg38 genome

```python
! wget https://github.com/hasindu2008/seq/raw/refs/heads/main/data/demo/hg38_chr22.fa
```

Perform the alignment to generate a SAM file and see if the output looks good

```python
! minimap2-2.28_x64-linux/minimap2 -a -x map-ont --secondary=no hg38_chr22.fa reads.fastq > reads.sam
```

```python
! head reads.sam
```

Install samtools, then convert SAM to its binary format BAM and index it

```python
! sudo apt install samtools
```

```python
! samtools sort reads.sam -o reads.bam
```

```python
! samtools index reads.bam
```

## Visualising the pileup of signals aligned against a reference

Run f5c eventalign to map the raw signals to the reference

```python
! f5c-v1.5/f5c_x86_64_linux_cuda index --slow5 chr22_20240000_20290000.blow5 reads.fastq
```

```python
! f5c-v1.5/f5c_x86_64_linux_cuda eventalign --slow5 chr22_20240000_20290000.blow5 -r reads.fastq -b reads.bam -a -g hg38_chr22.fa > eventalign.sam
```

Convert the eventalign SAM file from f5c to BAM and index

```python
! samtools sort eventalign.sam -o eventalign.bam && samtools index eventalign.bam
```

Now plot using squigualiser

```python
! squigualiser/squigualiser plot_pileup -f hg38_chr22.fa -s chr22_20240000_20290000.blow5 -a eventalign.bam -o pileup --region chr22:20271000-20272000
```

## Additional example: Downloading a few reads from a large remote BLOW5 file

The above tiny example BLOW5 we used had only ~100 reads. A full dataset is much more bigger than this. For example, see the size of the complete dataset that was used to extract the reads in the example: https://gtgseq.s3.amazonaws.com/ont-r10-5khz-dna/NA24385_e821/raw/PGXXSX240470_reads.blow5. It is one terabyte.

Now let us see how we can use slow5curl to selectively download a few random reads from this large terabyte file, without downloading the whole thing.


Download and extract slow5curl

```python
! wget https://github.com/BonsonW/slow5curl/releases/download/v0.2.1/slow5curl-v0.2.1-x86_64-linux-binaries.tar.gz && tar xf slow5curl-v0.2.1-x86_64-linux-binaries.tar.gz
```

Now let us print the read IDs of the first 1000 reads in the BLOW5 file into a list

```python
! slow5curl-v0.2.1/slow5curl reads https://gtgseq.s3.amazonaws.com/ont-r10-5khz-dna/NA24385_e821/raw/PGXXSX240470_reads.blow5 | head -1000 > list.txt
```

Now extract this list of reads from the big BLOW5

```python
! slow5curl-v0.2.1/slow5curl get https://gtgseq.s3.amazonaws.com/ont-r10-5khz-dna/NA24385_e821/raw/PGXXSX240470_reads.blow5 --list list.txt -o first1000reads.blow5
```

See if the extracted dataset is good

```python
! slow5tools-v1.3.0/slow5tools stats first1000reads.blow5
```
